 # Exercise: Complex Number Geometry

Goal: Represent complex numbers in the plane and visualize addition/multiplication.

The sume and product of two complex numbers:
$$z1 = z1 + z2$$
$$z2 = z1 * z2$$

Is as follows:

- $z_1 = a_1 + jb_1$
- $z_2 = a_2 + jb_2$
$$z_1 + z_2 = (a_1 + a_2) + j(b_1 + b_2)$$
$$z_1 \cdot z_2 = a_1a_2 + ja_1b_2 + ja_2b_1 - b_1b_2 = (a_1a_2 - b_1b_2) + j(a_1b_2 + a_2b_1)$$

In the complex plane, we easily see that, given $z = a + jb$, passing in polar coordinates:
- $\rho = \sqrt{a^2 + b^2}$
- $a = \rho \cos \theta$
- $b = \rho \sin \theta$
- $b/a = \tan \theta$ 

Modulus of z:
$$|z| =  \sqrt{a^2 + b^2}$$

Argument of z:
$$arg(\theta) = \arctan(b/a)$$

> $(+ \pi)$ iff b < 0

> In python write numbers in complex format like z1 = a + jb.

> The type of z1 is `complex`

if $z_1 = a + bj$ and $z_2 = a-bj$ $z_1$ and $z_2$ are conjugates, it is written $z_2 = \bar{z}_1$. 
From definition of modulus and argument:
$$z1 * \bar{z}_1 = |z1|^2$$
$$arg(z_1) = -arg (\bar{z}_1)$$


```python
import cmath

# Define two complex conjugates
z1 = 1 + 2j
z2 = z1.conjugate()
print("z1, z2", z1, z2)

# Using cmath module
p1 = cmath.phase(z1)
p2 = cmath.phase(z2)
print("arg(z1), arg(z2)", p1, p2)


# Plotting the points
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
plt.plot(z1.real, z1.imag, 'ro')  # Red dot for z1
plt.plot(z2.real, z2.imag, 'bo')  # Blue dot for z2

# Adding labels
# plt.text(x, y, "label_text")
plt.text(z1.real, z1.imag, f'  z1 ({z1})', verticalalignment='bottom')
plt.text(z2.real, z2.imag, f'  z2 ({z2})', verticalalignment='bottom')

# Add grid and labels for clarity
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.title('Complex Conjugates Numbers Visualization')

plt.show()
```
