 # Exercise: Complex Number Geometry

Goal: Represent complex numbers in the plane and visualize addition/multiplication.

The sume and product of two complex numbers:
$$z1 = z1 + z2$$
$$z2 = z1 * z2$$

Is as follows:

- z1 = a1 + jb1 
- z2 = a2 + jb2 
- $z1 + z2 = (a1 + a2) + j(b1 + b2)$ 
- $z1 * z2$ = a1a2 + ja1b2 +ja2b1 -b1b2 = $(a1a2 -b1b2) +j(a1b2 + a2b1)$

> In python write numbers in complex format like z1 = a + jb.

> type(z1) = complex

```python
# Define two complex numbers
z1 = 2 + 1j
z2 = -1 + 2j
# we expect z_sum = 1 + 3j

# Perform operations
z_sum = z1 + z2
z_prod = z1 * z2

# Plotting the points
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
plt.plot(z1.real, z1.imag, 'ro')  # Red dot for z1
plt.plot(z2.real, z2.imag, 'bo')  # Blue dot for z2
plt.plot(z_sum.real, z_sum.imag, 'go')  # Green dot for z_sum
plt.plot(z_prod.real, z_prod.imag, 'yo')  # Yellow dot for z_sum

# Adding labels
# plt.text(x, y, "label_text")
plt.text(z1.real, z1.imag, f'  z1 ({z1})', verticalalignment='bottom')
plt.text(z2.real, z2.imag, f'  z2 ({z2})', verticalalignment='bottom')
plt.text(z_sum.real, z_sum.imag, f'  z_sum ({z_sum})', verticalalignment='bottom')
plt.text(z_prod.real, z_prod.imag, f'  z_prod ({z_prod})', verticalalignment='bottom')

# Add grid and labels for clarity
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.title('Complex Numbers Visualization')

plt.show()
```
