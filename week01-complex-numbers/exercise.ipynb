Goal: Represent complex numbers in the plane and visualize addition/multiplication.

The sume and product of two complex numbers:
$$z1 = z1 + z2$$
$$z2 = z1 * z2$$

Is as follows:

- $z_1 = a_1 + jb_1$
- $z_2 = a_2 + jb_2$
$$z_1 + z_2 = (a_1 + a_2) + j(b_1 + b_2)$$
$$z_1 \cdot z_2 = a_1a_2 + ja_1b_2 + ja_2b_1 - b_1b_2 = (a_1a_2 - b_1b_2) + j(a_1b_2 + a_2b_1)$$

In the complex plane, we easily see that, given $z = a + jb$, passing in polar coordinates:
- $\rho = \sqrt{a^2 + b^2}$
- $a = \rho \cos \theta$
- $b = \rho \sin \theta$
- $b/a = \tan \theta$ 

Modulus of z:
$$|z| =  \sqrt{a^2 + b^2}$$

Argument of z (also said phase):
$$arg(\theta) = \arctan(b/a)$$

> $(+ \pi)$ iff b < 0

> In python write numbers in complex format like z1 = a + jb.

> The type of z1 is `complex`

if $z_1 = a + bj$ and $z_2 = a-bj$ $z_1$ and $z_2$ are conjugates, it is written $z_2 = \bar{z}_1$. 
From definition of modulus and argument:
$$z1 * \bar{z}_1 = |z1|^2$$
$$arg(z_1) = -arg (\bar{z}_1)$$


```python
import cmath

# Define two complex conjugates
z1 = 1 + 2j
z2 = z1.conjugate()
print("z1, z2", z1, z2)

# Using cmath module
p1 = cmath.phase(z1)
p2 = cmath.phase(z2)
print("arg(z1), arg(z2)", p1, p2)


# Plotting the points
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
plt.plot(z1.real, z1.imag, 'ro')  # Red dot for z1
plt.plot(z2.real, z2.imag, 'bo')  # Blue dot for z2

# Adding labels
# plt.text(x, y, "label_text")
plt.text(z1.real, z1.imag, f'  z1 ({z1})', verticalalignment='bottom')
plt.text(z2.real, z2.imag, f'  z2 ({z2})', verticalalignment='bottom')

# Add grid and labels for clarity
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.title('Complex Conjugates Numbers Visualization')

plt.show()
```

The addtion and multiplication of complex numbers has a geometric representation

**Addition** of two vectors is as the sum of two vectors using the vector sum rule.

**Multiplication** of two vectors of same modulus is like a rotation (modulus is doubled).

For example $(1 + j) +  (2 + j) = (3 + 2j)$

For example $(1 + j) * (1 + j) = (1 -1 +2j)^2 = 2j$  

See graph below.

```python
import cmath
import matplotlib.pyplot as plt
import numpy as np

# Define complex numbers
z1 = 1 + 1j
z2 = 2 + 1j
z3 = z1 + z2
z4 = z1 * z1

# Setup plot
plt.figure(figsize=(8, 8))
ax = plt.gca()
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)

# Define a helper list to iterate through for cleaner code
points = [
    (z1, 'ro', 'red', 'z1'),
    (z2, 'bo', 'blue', 'z2'),
    (z3, 'go', 'green', 'z1 + z2'),
    (z4, 'yo', 'yellow', 'z1 * z1')
]

for z, marker, color, label in points:
    # 1. Draw the vector (arrow) from (0,0) to z
    # angles='xy' ensures the arrow points to the correct coordinate
    # scale_units='xy' and scale=1 ensure the arrow length matches the axes
    plt.quiver(0, 0, z.real, z.imag, angles='xy', scale_units='xy', 
                scale=1, color=color, alpha=0.6, label=label)
    
    # 2. Plot the point (dot)
    plt.plot(z.real, z.imag, marker)
    
    # 3. Add labels
    plt.text(z.real, z.imag, f'  {label} ({z})', verticalalignment='bottom')

# Add grid and labels for clarity
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.title('Complex Numbers Visualization with Vectors')

plt.show()
```

> `plt.quiver(0, 0, z.real, z.imag, ...)`

- The first two arguments (0, 0) are the starting point (the origin).
- The next two arguments (z.real, z.imag) are the direction and magnitude of the vector.

- angles='xy' 
> This is crucial. By default, quiver uses the "data" angle. Setting it to 'xy' ensures that the arrow points exactly to the $(x, y)$ coordinate on your plot.

- scale_units='xy' and scale=1: 
> These ensure that the length of the arrow is proportional to the numbers on your axes. Without these, Matplotlib might "shrink" or "stretch" the arrows based on how many vectors are on the screen.

- alpha=0.6: 
> added a slight transparency to the arrows so that if they overlap, you can still see both.
