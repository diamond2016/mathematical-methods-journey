**Euler's formula** establishes a fundamental bridge between trigonometric analysis and exponential analysis through the identity:


$$e^{i\theta} = \cos\theta + i\sin\theta$$


for any real angle $\theta$.

This connection relates directly to the **polar form** and the **calculation of complex roots** in the following ways:

1. **Compact Polar Representation:** Any non-zero complex number $z$ can be written in trigonometric form as $z = r(\cos\theta + i\sin\theta)$. By utilizing Euler's formula, this expression takes on an extremely compact exponential form:



$$z = r e^{i\theta}$$



where $r = \vert{}z\vert{}$ is the modulus and $\theta = \arg(z)$ is the argument.


2. **Simplification of Powers and Roots (De Moivre's Theorem / Properties of Exponents):** When calculating powers or extracting the $n$-th roots of a complex number, exponential notation makes the algebraic steps straightforward, thanks to the laws of exponents ($e^{a} \cdot e^{b} = e^{a+b}$ and $(e^{a})^b = e^{a \cdot b}$).
3. **Systematic Calculation of $n$-th Roots:** To solve the equation $w^n = z$, where $z = r e^{i\theta}$, we express the root we are looking for in exponential form: $w = R e^{i\varphi}$. Raising to the $n$-th power, we obtain:



$$w^n = R^n e^{i n\varphi} = r e^{i\theta}$$



Since the argument of a complex number is periodic (it can vary by integer multiples of $2\pi$), we set $n\varphi = \theta + 2k\pi$ (where $k \in \mathbb{Z}$). From this, the $n$ distinct roots follow directly by applying Euler's formula to each angle:



$$w_k = \sqrt[n]{r} \, e^{i \frac{\theta + 2k\pi}{n}} = \sqrt[n]{r} \left( \cos \frac{\theta + 2k\pi}{n} + i\sin \frac{\theta + 2k\pi}{n} \right)$$



for $k = 0, 1, 2, \dots, n-1$.
