# Analysis of Complex Transformations

## Why It Matters

Transformations in the complex plane, such as the mapping $w = 1/z$, connect Euclidean geometry to deep algebraic properties. Understanding how to algebraically verify the radius and geometric behavior allows you to develop rigorous metacognitive checks and avoid critical errors in the study of complex functions.

## Key Concepts

### Geometry of Inversion (1/z)

The transformation $w = 1/z$ maps circles and lines into other circles or lines. Geometrically, inversion with respect to the unit circle inverts the radius into its reciprocal and reflects the point across the real axis.

1. **Express the complex number in polar form:**
Write $z = r e^{i\theta}$, where $r$ is the modulus and $\theta$ is the argument.


2. **Calculate the reciprocal:**
Apply the definition: $w = \frac{1}{r e^{i\theta}} = \frac{1}{r} e^{-i\theta}$.


3. **Interpret the geometric result:**
The modulus of the new point is $1/r$ and the argument is $-\theta$ (angular reflection).


### Algebraic Verification of the Radius Parameter

To ensure that the algebraic parameters match the expected geometric behavior, we use the Cartesian representation and the properties of complex moduli.

## Worked Examples

Let's now test the concept by applying the algebraic radius verification to concrete cases of geometric transformation.

**Example 1: Image of a circle of radius R**
Determine the equation of the image of the circle $\vert{}z\vert{} = R$ via the transformation $w = 1/z$.

1. **Set up the algebraic relation:**
Let $z$ be a point on the circle, so $\vert{}z\vert{} = R$. We know that $w = 1/z$, hence $z = 1/w$.


2. **Substitute to find the modulus of w:**
Calculate the modulus of $w$: $\vert{}w\vert{} = \left\vert{} \frac{1}{z} \right\vert{} = \frac{1}{\vert{}z\vert{}}$.


3. **Conclude with the radius verification:**
Since $\vert{}z\vert{} = R$, we obtain $\vert{}w\vert{} = \frac{1}{R}$. The image is a circle of radius $1/R$.


**Example 2: Verification on a specific point**
Verify the effect of the transformation on the point $z = 2i$.

1. **Calculate the initial modulus:**
The point has modulus $\vert{}2i\vert{} = 2$.


2. **Apply the transformation:**
$w = \frac{1}{2i} = \frac{-i}{2} = -\frac{1}{2}i$.


3. **Verify the final radius:**
The modulus of the transformed point is $\vert{}w\vert{} = \left\vert{}-\frac{1}{2}i\right\vert{} = \frac{1}{2}$, confirming that $1/2 = 1/\vert{}z\vert{}$.
