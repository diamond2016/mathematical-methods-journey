As stated Cauchy - Riemann (CR) equations allow to check if a complex function, expressed as $f(z = x + iy) = u(x, y) + iv(x,y)$ is analytic (or holomorphic).

*Holomorphic Function* (holomorphic):

Meaning: Derived from the Greek *holos* (whole) and *morphe* (form). It describes a function that possesses a complex derivative at every point within an open neighborhood (i.e., it is complex-differentiable).

*Analytic Function* (analytic):

Meaning: Refers to a function that can be expressed as a convergent power series (Taylor series) at every point within its domain.

*A complex function is holomorphic on an open domain if and only if it is analytic on that domain*.

The program `cauchy_riemann_check.py` make check of analyticity using CR condition.

Note: we use j as imaginary unit.

*Exercise 1*

Is the function $$f(z) = z^2$$ analytic (holomorphic) in all the complex plane?

1. Write $f(z)$ in terms of the Cartesian coordinates $x$ and $y$, remembering that $z = x + iy$. 

> $f(z) = x + jy = (x + jy)^2 = x^2 -y^2 +2jxy$

2. Separate the real part $u(x,y)$ and the imaginary part $v(x,y)$, so that $f(x+iy) = u(x,y) + i v(x,y)$. 

> * $u(x, y) = x^2 -y^2$
> * $v(x, y) = 2xy$

3. Compute the first-order partial derivatives: 
* $\frac{\partial u}{\partial x}$ and $\frac{\partial v}{\partial y}$
* $\frac{\partial u}{\partial y}$ and $\frac{\partial v}{\partial x}$

> $\frac{\partial u}{\partial x} = 2x$ and $\frac{\partial v}{\partial y}$ = 2x
> $\frac{\partial u}{\partial y}$ = -2y$ and $\frac{\partial v}{\partial x}$ = 2y

4. Checks whether the Cauchy-Riemann conditions hold:

> $\frac{\partial u}{\partial x} = 2x = \frac{\partial v}{\partial y} = 2x$ - holds;
> $\frac{\partial u}{\partial y} = -2y  = -\frac{\partial v}{\partial x} 2y$ - holds.

So f(z) is analytic (holomorphic).
